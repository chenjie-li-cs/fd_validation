```python
from ucimlrepo import fetch_ucirepo 

def download_full_dataset_as_df(did, dname):
    
    # fetch dataset 
    dataset = fetch_ucirepo(id=did) 
      
    df = dataset.data.features
    df['target'] = dataset.data.targets  # If there's a target variable
    df.to_csv(f'uci_{dname}.csv',index=False)
    # return df
```

```python
download_full_dataset_as_df(did=12, dname='balance_scale')
```
