```python
def plot_res(dfs):
    for df in dfs:
        measure = df['measure'][0]  # Get the measure name
        plt.figure(figsize=(8, 5))
        plt.bar(df['fd'], df['val'], color='blue', alpha=0.7)
        plt.xlabel('Functional Dependency')
        plt.ylabel('Value')
        plt.title(f'Measure: {measure}')
        plt.xticks(rotation=45)
        plt.ylim(0, 1.05)
        plt.show()
```
