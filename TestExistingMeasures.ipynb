```python
def test_general_and_original(func_name, original_func, general_func, lhs, rhs, df):
    print(f"{func_name}")
    print(original_func(df=df, lhs=lhs[0], rhs=rhs[0]))
    print(general_func(df=df, lhs=lhs, rhs=rhs))
    assert original_func(df=df, lhs=lhs[0], rhs=rhs[0]) == general_func(df=df, lhs=lhs, rhs=rhs), f"mismatch in {func_name}"
```

```python
d = {
    'g1':[g1, g1_general],
    'g2':[g2, g2_general],
    'g3':[g3, g3_general],
    'g3_prime': [g3_prime, g3_prime_general],
    'g1_prime': [g1_prime, g1_prime_general],
    'rho': [rho, rho_general],
    'shannon_g1': [shannon_g1, shannon_g1_general],
    'shannon_g1_plus': [shannon_g1_plus, shannon_g1_plus_general],
    'FI': [fraction_of_information, fraction_of_information_general],
    'RFI': [reliable_fraction_of_information, reliable_fraction_of_information_general],
    'RFI_plus': [reliable_fraction_of_information_plus,reliable_fraction_of_information_plus_general],
    'pdep': [pdep, pdep_general],
    'tau':[tau, tau_general],
    'mu':[mu, mu_general],
    'mu_plus': [mu_plus, mu_plus_general]
}
```

```python
generalized_measures = {
'g1': g1_general,
'g2': g2_general,
'g3': g3_general,
'g3_prime': g3_prime_general,
'g1_prime': g1_prime_general,
'rho': rho_general,
'shannon_g1': shannon_g1_general,
'shannon_g1_plus': shannon_g1_plus_general,
'fraction_of_information': fraction_of_information_general,
'reliable_fraction_of_information': reliable_fraction_of_information_general,
'reliable_fraction_of_information_plus': reliable_fraction_of_information_plus_general,
'pdep': pdep_general,
'tau': tau_general,
'mu': mu_general,
'mu_plus': mu_plus_general
}
```
